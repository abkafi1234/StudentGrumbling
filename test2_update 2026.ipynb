```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

# 1. LOAD AND CLEAN DATA
# Replace the filename with your actual file path
file_path = './GRUMBLING BEHAVIOR (STEM) (Responses).xlsx' 
df = pd.read_excel(file_path)

# Map Likert responses to integers
likert_map = {
    'Strongly disagree': 1, 'Disagree': 2, 'Neutral': 3, 
    'Agree': 4, 'Strongly agree': 5
}

# Define your columns based on your survey order
# (Adjust these indices if your column order is different)
cols = {
    'SC': [df.columns[4], df.columns[5]],   # Switching Cost
    'SB': [df.columns[6], df.columns[7]],   # Switching Benefit
    'WO': [df.columns[8], df.columns[9]],   # Work Overload
    'RA': [df.columns[10], df.columns[11]], # Role Ambiguity
    'WHC': [df.columns[12]],                # Work-Home Conflict (Single Item)
    'GG': [df.columns[13], df.columns[14], df.columns[15]], # General Grumbling
    'W': [df.columns[16], df.columns[17]],  # Willingness
    'LTA': [df.columns[18]]                 # Long Term Adoption (Single Item)
}

# Convert and Create Composite Scores (The "Fix")
df_path = pd.DataFrame()
for construct, item_list in cols.items():
    temp_df = df[item_list].replace(likert_map)
    df_path[construct] = temp_df.mean(axis=1)

# 2. PATH ANALYSIS (STEP-BY-STEP MEDIATION)
print("--- PATH 1: Drivers of Work-Home Conflict ---")
model_whc = smf.ols('WHC ~ WO + SC', data=df_path).fit()
print(model_whc.summary())

print("\n--- PATH 2: Drivers of Willingness ---")
model_w = smf.ols('W ~ SB', data=df_path).fit()
print(model_w.summary())

print("\n--- PATH 3: THE MAIN EVENT - Drivers of General Grumbling ---")
# This tests what actually causes the resistance
model_gg = smf.ols('GG ~ RA + WHC + SC + SB + W + LTA + WO', data=df_path).fit()
print(model_gg.summary())

# 3. CHECK FOR REDUNDANCY (Multicollinearity)
from statsmodels.stats.outliers_influence import variance_inflation_factor
X = df_path[['RA', 'WHC', 'SC', 'SB', 'W', 'LTA', 'WO']]
X = sm.add_constant(X)
vif = pd.DataFrame()
vif["Variable"] = X.columns
vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print("\n--- MULTICOLLINEARITY CHECK (VIF) ---")
print(vif)
```
